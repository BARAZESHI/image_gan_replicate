```python
import replicate
import requests
import json
from PIL import Image
from io import BytesIO
```

```python
import gradio as gr

def greet(prompt):
    output = replicate.run(
    "google/imagen-4",
    input={
        "prompt": f"{prompt}",
        "aspect_ratio": "16:9",
        "output_format": "jpg",
        "safety_filter_level": "block_medium_and_above"

    }
)

    url = output.url

    response = requests.get(url)
    img = Image.open(BytesIO(response.content))
    return img

demo = gr.Interface(
    fn=greet,
    inputs=["text"],
    outputs=["image"],
    api_name="predict"
)

demo.launch()
```
